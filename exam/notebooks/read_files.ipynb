```python
import pandas as pd
import numpy as np
```

```python
columns = [
    "age", "sex", "cp", "trestbps", "chol", "fbs",
    "restecg", "thalach", "exang", "oldpeak",
    "slope", "ca", "thal", "num"
]
```

```python
df_cleveland = pd.read_csv("../data/processed.cleveland.data",
                            names = columns,
                            na_values = "?")

df_hungarian = pd.read_csv("../data/processed.hungarian.data",
                            names = columns,
                            na_values = "?")

df_swiss = pd.read_csv("../data/processed.switzerland.data",
                            names = columns,
                            na_values = "?")

df_va = pd.read_csv("../data/processed.va.data",
                            names = columns,
                            na_values = "?")

# add origin column
df_cleveland["origin"] = "CLE"
df_hungarian["origin"] = "HUN"
df_swiss["origin"] = "SW"
df_va["origin"] = "VA"
```

```python
print(df_cleveland.shape)
print(df_cleveland.isna().sum())
```

```python
print(df_hungarian.shape)
print(df_hungarian.isna().sum())
df_hungarian["num"].value_counts()
```

```python
print(df_swiss.shape)
print(df_swiss.isna().sum())
```

```python
print(df_va.shape)
print(df_va.isna().sum())
```

```python
df_full = pd.concat([df_cleveland, df_hungarian, df_swiss, df_va])
# encode the response to binary
df_full["target"] = (df_full["num"] > 0).astype(int)
# df_full = df_full.drop(columns=["num"])
df_full.columns
```

```python
df_dropped = df_full.dropna()
df_dropped.shape
```

```python
df_partial = df_full[[col for col in df_full.columns if col not in ["slope", "ca", "thal"]]]
df_partial = df_partial.dropna()
df_partial.shape
```

```python
df_rich = df_dropped
df_large = df_partial
```

```python
print(df_rich.shape)
print(df_large.shape)
print(df_rich.columns)
print(df_large.columns)
```
