```python
import math
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

```

```python
data = [
    {"customer_id": 1, "income_raw": "45000",   "spending": 800},
    {"customer_id": 2, "income_raw": ">50000",  "spending": 1200},
    {"customer_id": 3, "income_raw": "32000",   "spending": 500},
    {"customer_id": 4, "income_raw": "28000",   "spending": 400},
    {"customer_id": 5, "income_raw": ">50000",  "spending": 1500},
]
df = pd.DataFrame(data)
df
```

```python
def parse_income(value: str):
    value = str(value).strip().replace(",", "")
    if value.startswith(">"):
        return float(value.replace(">", "")), True
    return float(value), False
```

```python
parsed = df["income_raw"].apply(parse_income)
df["income_observed"] = parsed.apply(lambda x: x[0])
df["is_censored"] = parsed.apply(lambda x: x[1])

print("Parsed dataset:")
print(df)
```

Initialization

```python
observed = df["income_observed"].to_numpy()

mu_init = observed.mean()
sigma2_init = sum([(i-mu_init)**2 for i in observed])/len(observed)
```

E-Step

```python
def std_norm_pdf(x: float) -> float:
    return math.exp(-0.5 * x * x) / math.sqrt(2.0 * math.pi)

def std_norm_cdf(x: float) -> float:
    return 0.5 * (1.0 + math.erf(x / math.sqrt(2.0)))
```
